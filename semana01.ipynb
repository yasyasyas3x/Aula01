```python
import pandas as pd
df = pd.read_excel('seu_arquivo.xlsx')
display(df.head(6))
display(df.tail(6))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Nome da coluna a ser analisada
column_name = "Qual dessas frutas você prefere para suco natural?"

# Contar a frequência de cada fruta
fruit_counts = df[column_name].value_counts().reset_index()
fruit_counts.columns = ['Fruta', 'Contagem']

# Criar o gráfico de barras
plt.figure(figsize=(10, 5))
sns.barplot(x='Fruta', y='Contagem', data=fruit_counts, palette='viridis')
plt.title('Preferência de Frutas para Suco Natural')
plt.xlabel('Fruta')
plt.ylabel('Número de Respostas')
plt.xticks(rotation=360, ha='right') # Rotaciona os rótulos para melhor leitura
plt.tight_layout() # Ajusta o layout para evitar cortes
plt.show()
```
